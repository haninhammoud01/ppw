# Business Understanding

## 1. Define the Problem

```{admonition} Ringkasan Masalah
:class: note

Mengumpulkan **200 artikel berita Indonesia** dari [detik.com](https://detik.com) sebagai dataset untuk tugas mata kuliah Pencarian dan Penambangan Web (PPW), disusun mengikuti metodologi CRISP-DM. Fokus masalahnya adalah kategorisasi berita Indonesia (sport vs finance), yang akan dikerjakan pada tahap-tahap CRISP-DM selanjutnya.
```

## 2. Set Objectives

```{admonition} Tujuan
:class: tip

- Membangun dataset berlabel yang bersih berisi 200 artikel (100 **sport**, 100 **finance**) dengan skema kolom `id`, `isi_berita`, `label`, `url`.
- Menyiapkan data agar tahap CRISP-DM berikutnya (Data Preparation, Modeling, Evaluation, Deployment) bisa dikerjakan pada tugas-tugas selanjutnya.
```

## 4. Define Success Criteria

```{admonition} Kriteria Keberhasilan
:class: important

- Tepat **200 baris**: 100 `sport` (id 1-100) dan 100 `finance` (id 101-200)
- **0 nilai kosong** pada `isi_berita`; `id` berurutan 1-200
- Dataset dapat dimuat tanpa error dan sesuai skema kolom yang ditentukan (`id`, `isi_berita`, `label`, `url`)
```

## 5. Assess Situation (Constraints & Ethics)

```{admonition} Etika & Batasan Crawling
:class: warning

Proses pengumpulan data memperhatikan crawling yang sopan dan bertanggung jawab:

- **Politeness**: setiap request diberi jeda minimal 1 detik dan menggunakan header User-Agent seperti browser, agar tidak membebani server.
- **Tujuan**: data hanya digunakan untuk **keperluan pembelajaran / tugas mata kuliah ini**.
- **Sumber data**: seluruh URL artikel diambil secara otomatis dari halaman indeks kategori resmi Detik.com (`sport.detik.com/indeks`, `finance.detik.com/indeks`), bukan dikumpulkan/ditulis manual.
```
